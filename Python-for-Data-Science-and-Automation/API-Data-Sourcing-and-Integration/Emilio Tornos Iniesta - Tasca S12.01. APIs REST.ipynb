### Exercicis 1 i 2
1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:
   * Llista de publicacions (/posts)
   * Llista d’usuaris (/users)
   * Llista de tasques (/todos)
2. Mostra per pantalla:
   * La quantitat total de cada recurs.
   * El codi d'estat de cada petició.

```python
# Defino las URLs a las que voy a llamar
urlPosts = "https://jsonplaceholder.typicode.com/posts"
urlUsers = "https://jsonplaceholder.typicode.com/users"
urlTodos = "https://jsonplaceholder.typicode.com/todos"

# Hago las peticiones GET
respPosts = requests.get(urlPosts)
respUsers = requests.get(urlUsers)
respTodos = requests.get(urlTodos)

# Extraigo los datos en formato JSON (me lo convierte a listas/diccionarios de Python)
datosPosts = respPosts.json()
datosUsers = respUsers.json()
datosTodos = respTodos.json()

# Imprimo los resultados
print("CONSULTA (GET)")
print(f"Posts -> Código de estado: {respPosts.status_code} | Cantidad total: {len(datosPosts)}")
print(f"Users -> Código de estado: {respUsers.status_code} | Cantidad total: {len(datosUsers)}")
print(f"Todos -> Código de estado: {respTodos.status_code} | Cantidad total: {len(datosTodos)}")
```

### Exercici 4
Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.
* Mostra la resposta JSON.
* Mostra el codi d'estat.

```python
# CREAR RECURSO (POST)
urlCrear = "https://jsonplaceholder.typicode.com/posts"

# Preparo los datos ficticios que voy a enviar
nouPost = {
    "title": "Mi título para el ejercicio",
    "body": "Este es el cuerpo del mensaje para aprobar el SPRINT 12",
    "userId": 12
}

# Hago la petición POST enviando mi diccionario
respPost = requests.post(urlCrear, json=nouPost)

print("CREAR PUBLICACIÓN (POST)")
print(f"Código de estado POST: {respPost.status_code}") # Suele devolver 201 (Created)
print("Respuesta JSON:", respPost.json())
```

### Exercici 5
Fes una petició PATCH per modificar parcialment una publicació existent.
* Mostra la resposta JSON.
* Mostra el codi d'estat.

```python
# MODIFICAR RECURSO (PATCH)
# Voy a modificar el post número 1, por eso pongo /posts/1
urlModificar = "https://jsonplaceholder.typicode.com/posts/1"

# Solo quiero cambiar el título, el resto de campos se quedarán igual
cambioTitulo = {
    "title": "Título modificado con PATCH por Emilio"
}

# Lanzo el PATCH
respPatch = requests.patch(urlModificar, json=cambioTitulo)

print("MODIFICAR PUBLICACIÓN (PATCH)")
print(f"Código de estado PATCH: {respPatch.status_code}") # Devolverá 200 (OK)
print("Respuesta JSON:", respPatch.json())
```

### Exercici 6
Fes una petició DELETE sobre una publicació.
* Mostra la resposta JSON.
* Mostra el codi d'estat.

```python
# BORRAR RECURSO (DELETE)
# Voy a borrar el post número 1
urlBorrar = "https://jsonplaceholder.typicode.com/posts/1"

# Lanzo el DELETE
respDelete = requests.delete(urlBorrar)

print("BORRAR PUBLICACIÓN (DELETE)")
print(f"Código de estado DELETE: {respDelete.status_code}") # Devolverá 200 (OK)
print("Respuesta JSON:", respDelete.json()) # Suele devolver un diccionario vacío {}
```

### Ejercicios 1 y 2: Exploración de una API Pública (Star Wars)

He elegido la **SWAPI (The Star Wars API)** (https://swapi.dev/) porque es un estándar que considero genial para practicar: es gratuita, no requiere autenticación y su documentación es muy limpia.

**Endpoints:**
Al leer la documentación, he encontrado que la API tiene varios endpoints organizados por temática. Me anoto estos dos:
1. `/people`: Para obtener la información de los personajes (altura, peso, color de ojos, etc.).
2. `/planets`: Para consultar los planetas del universo Star Wars y sus climas.

**Filtros y parámetros opcionales:**
Me fijo en que la API permite filtrar y buscar datos usando parámetros en la URL (query parameters). Por ejemplo, en el endpoint `/people` puedo usar:
* `search`: Para buscar personajes directamente por su nombre o parte de él (ej: `/?search=skywalker`).
* `page`: Para navegar por los resultados si hay muchos, ya que vienen paginados (ej: `/?page=2`).

**Formato JSON:**
He verificado que la respuesta siempre viene en formato JSON, estructurado con una clave principal llamada `results` donde se guardan los datos, perfecto para Pandas.

### Exercici 3
Fes una petició GET senzilla:
* Mostra el codi d'estat de la resposta.
* Imprimeix de forma clara alguns dels camps de la resposta JSON.

```python
# Defino la URL de la API aplicando el filtro de búsqueda que mencioné arriba
urlApi = "https://swapi.dev/api/people/?search=skywalker"

# Hago la petición GET
respStarWars = requests.get(urlApi)

print("PETICIÓN GET")
print(f"Código de estado: {respStarWars.status_code}") # Me tiene que devolver un 200 (OK)

# Convierto la respuesta a formato JSON
datosJson = respStarWars.json()

# La API guarda la lista de los personajes dentro de la clave "results"
listaPersonajes = datosJson["results"]

print("\nImprimo algunos campos de los personajes encontrados:")
for personaje in listaPersonajes:
    # Navego por el JSON para sacar el nombre, la altura y el peso
    print(f"- Nombre: {personaje['name']} | Altura: {personaje['height']} cm | Peso: {personaje['mass']} kg")
```

### Exercici 4
Converteix les dades a un DataFrame de pandas:
* Mostra les primeres files del DataFrame.

```python
# Convierto mi lista de diccionarios directamente a un DataFrame
dfStarWars = pd.DataFrame(listaPersonajes)

# Selecciono solo las columnas que me interesan para que quede una tabla limpia
columnasInteres = ["name", "height", "mass", "hair_color", "eye_color", "gender"]
dfLimpio = dfStarWars[columnasInteres]

print("Muestro las primeras filas de mi tabla:")

# Muestro la cabecera de la tabla para confirmar que se ha creado bien
dfLimpio.head()
```

# Nivell 3
API d'Open Data Barcelona

### Exercicis 1 i 2
1. Utilitza l’API d’Open Data BCN per: Cercar un dataset del teu interès mitjançant `package_search` i `package_show`.
2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

```python
# Defino la URL base de la API del Ayuntamiento de Barcelona
urlBaseBcn = "https://opendata-ajuntament.barcelona.cat/data/api/3/action/"

# BUSCAR DATASET (package_search)
# Busco un paquete de datos relacionado con el Wi-Fi público
urlBuscar = urlBaseBcn + "package_search?q=wifi"

# Hago la petición GET
respBuscar = requests.get(urlBuscar)
datosBuscar = respBuscar.json()

# Me quedo con el primer dataset que me devuelve la búsqueda
primerDataset = datosBuscar["result"]["results"][0]
idDataset = primerDataset["id"]
tituloDataset = primerDataset["title"]

print("BUSCAR DATASET")
print(f"Dataset seleccionado: {tituloDataset}")

# SELECCIONAR RECURSO (package_show)
# Ahora miro qué archivos tiene dentro este dataset concreto
urlMostrar = urlBaseBcn + f"package_show?id={idDataset}"
respMostrar = requests.get(urlMostrar)
datosMostrar = respMostrar.json()

recursosDataset = datosMostrar["result"]["resources"]

# Recorro los archivos buscando uno que sea formato CSV
idRecursoElegido = None
for recurso in recursosDataset:
    if recurso["format"].upper() == "CSV":
        idRecursoElegido = recurso["id"]
        break

print("\nSELECCIONAR RECURSO CSV")
print(f"ID del archivo CSV encontrado: {idRecursoElegido}")
```

### Exercici 3
Amb el `resource_id` del recurs seleccionat, realitza una consulta mitjançant `datastore_search` per:
* Recuperar almenys 100 registres.

```python
# RECUPERAR REGISTROS (datastore_search)
# Le pido a la API que me traiga exactamente 100 registros usando el límite
urlDatos = urlBaseBcn + f"datastore_search?resource_id={idRecursoElegido}&limit=100"

# Hago la petición GET
respDatos = requests.get(urlDatos)
datosExtraidos = respDatos.json()

# Saco la lista de registros que está escondida dentro del JSON
listaRegistros = datosExtraidos["result"]["records"]

print("RECUPERAR REGISTROS")
print(f"Código de estado: {respDatos.status_code}") # Tiene que devolver 200 (OK)
print(f"Cantidad de registros recuperados: {len(listaRegistros)}")
```

### Exercicis 4 i 5
4. Converteix els resultats a un DataFrame.
5. Desa el DataFrame en un fitxer .csv.

```python
# CONVERTIR A DATAFRAME
# Le paso mi lista de diccionarios directamente a Pandas
dfWifi = pd.DataFrame(listaRegistros)

# GUARDAR EN CSV
# Guardo el archivo en mi ordenador sin la columna del índice numérico
nombreArchivo = "puntosWifiBcn.csv"
dfWifi.to_csv(nombreArchivo, index=False, encoding="utf-8")

print("CONVERTIR Y GUARDAR")
print(f"Archivo '{nombreArchivo}' generado con éxito en tu carpeta.")

# Muestro las primeras filas para asegurarme de que la tabla está perfecta
dfWifi.head()
```
