## 6.
* Crea una columna nova que sigui la data de naixament en format Datetime a partir de les columnes dia, mes i any. 
* Utilitzant aquesta columna crea una funció que donada una data, et calculi l'edat actual a dia d'avui.
- Utilitza la funció que acabes de crear per generar una columna nova al DataFrame amb l'edat actual.

```python
# Creo la columna "Data de Naixement" combinando las columnas de la tabla
# Renombro temporalmente las columnas para que Pandas las reconozca automáticamente
dfEncuesta["Data de Naixement"] = pd.to_datetime(dfEncuesta[["Dia", "Mes", "Any"]].rename(columns={
    "Dia": "day",
    "Mes": "month",
    "Any": "year"}))
```

```python
# Defino la función para calcular la edad exacta
def calcularEdad(dataNaixement):
    dataActual = datetime.now()
    # Resto los años
    edat = dataActual.year - dataNaixement.year
    # Compruebo si ya ha pasado el cumpleaños este año
    # Si el mes/día actual es menor al de nacimiento, resta 1
    haPassatAniversari = (dataActual.month, dataActual.day) >= (dataNaixement.month, dataNaixement.day)
    if not haPassatAniversari:
        edat -= 1
    return edat
```

```python
# Aplico la función para generar la nueva columna "Edat"
dfEncuesta["Edat"] = dfEncuesta["Data de Naixement"].apply(calcularEdad)
```

### La funció ha de crear (i exportar) un gràfic automàticament per a cada columna del dataframe. Per exemple:

* un histograma/boxplot si la variable és numèrica
* unes barres dels valors més freqüents si és categòrica
* unes barres dels anys més freqüents si la dada està en format data.

### Mostra el resultat de la funció en algun dels datasets d’exemple que conté el paquet seaborn. Per exemple, iris, penguins o titanic.

- Tingues en consideració que en el següent sprint treballaràs exclusivament amb gràfics. 
- L’objectiu d’aquest exercici no és crear gràfics molt elaborats, sinó resoldre una necessitat de manera ràpida i automàtica.



### Código de la función

```python
def generarGraficosAutomaticos(dfEntrada):
    # Recorre cada columna y genera un gráfico según su contenido.
    for columna in dfEntrada.columns:
        # Detecta el tipo de dato de la columna
        esNumerico = pd.api.types.is_numeric_dtype(dfEntrada[columna])
        esFecha = pd.api.types.is_datetime64_any_dtype(dfEntrada[columna])
        
        plt.figure(figsize=(8, 4))
        
        if esFecha:
            # Gráfico de barras de años más frecuentes
            años = dfEntrada[columna].dt.year.value_counts().sort_index()
            años.plot(kind="bar", color="skyblue")
            plt.title(f"Frecuencia por Año: {columna}")
        
        elif esNumerico:
            # Histograma para variables numéricas
            sns.histplot(dfEntrada[columna].dropna(), kde=True, color="salmon")
            plt.title(f"Distribución Numérica: {columna}")
        
        else:
            # Barras de valores más frecuentes para categorías
            topValors = dfEntrada[columna].value_counts().head(10)
            topValors.plot(kind="bar", color="lightgreen")
            plt.title(f"Top 10 Valores: {columna}")
            plt.xticks(rotation=45)

        plt.tight_layout()
        nombreArchivo = f"grafico{columna}.png"
        plt.savefig(nombreArchivo)
        plt.close()
        print(f"Archivo exportado: {nombreArchivo}")
```

* Carrega l'arxiu matriu_distancies.xlsx a pandas, de manera que els noms de files i els noms de columnes siguin els de les ciutats.

##### Ens interessa visitar totes les ciutats principals d'Espanya recorrent la mínima distància possible.

##### No cal que ho facis de forma òptima, ens interessa que desenvolupis una solució raonable utilitzant les eines que tens actualment.

##### Per exemple, una aproximació senzilla (que no òptima) seria anant sempre a la ciutat més propera que no haguem visitat encara

##### Fes una funció que donada la matriu de distàncies i la ciutat d'origen, faci una proposta de ruta que sigui el més curta possible que puguis, retornant una llista amb l'ordre de visita. Dóna també la distància total recorreguda.

##### (EXTRA) Des de quina ciutat la ruta seria més curta amb l'algoritme plantejat

### Código de la función

```python
# Cargo y limpio la matriz
dfDistancias = pd.read_excel("matriu_distancies.xlsx", index_col=0)
ciudadesABorrar = ["Las Palmas de Gran Canaria", "Palma"]
dfDistanciasLimpio = dfDistancias.drop(columns=ciudadesABorrar, errors="ignore").drop(index=ciudadesABorrar, errors="ignore")
dfDistanciasLimpio = dfDistanciasLimpio.fillna(0) # La distancia a uno mismo es 0

def calcularRutaMasCorta(matrizDistancias, ciudadOrigen, mostrarResultados=True):

    # Propuesta de ruta basada en el vecino más cercano.
    # Inicializo las listas necesarias para devolver un DataFrame con el desglose paso a paso
    ciudadesPendientes = list(matrizDistancias.index)
    ciudadesPendientes.remove(ciudadOrigen)
    
    listaOrden = [1]
    listaCiudades = [ciudadOrigen]
    listaDistanciaTramo = [0.0]
    listaDistanciaAcumulada = [0.0]
    
    distanciaTotal = 0.0
    ciudadActual = ciudadOrigen
    paso = 2
    
    while ciudadesPendientes:
        # Buscamos la distancia a las ciudades que quedan
        distanciasActuales = matrizDistancias.loc[ciudadActual, ciudadesPendientes]
        proximaCiudad = distanciasActuales.idxmin()
        distanciaTram = distanciasActuales.min()
        
        # Actualizamos la ruta
        distanciaTotal += distanciaTram
        
        # Registro los datos de cada tramo en mis listas para construir el informe
        listaOrden.append(paso)
        listaCiudades.append(proximaCiudad)
        listaDistanciaTramo.append(distanciaTram)
        listaDistanciaAcumulada.append(distanciaTotal)
        
        ciudadesPendientes.remove(proximaCiudad)
        ciudadActual = proximaCiudad
        paso += 1
        
    # Construyo el DataFrame final con toda la información recolectada
    dfResultadoRuta = pd.DataFrame({
        "Orden": listaOrden,
        "Ciudad": listaCiudades,
        "Distancia Tramo (km)": listaDistanciaTramo,
        "Distancia Acumulada (km)": listaDistanciaAcumulada})
    
    # Realizo las impresiones dentro de la función para que el usuario no tenga que gestionar los prints
    if mostrarResultados:
        # Primero convierto el DataFrame a string para medir su ancho real
        tablaTexto = dfResultadoRuta.to_string(index=False)
        anchoTabla = max(len(linea) for linea in tablaTexto.split('\n'))
        
        # Uso el ancho calculado para centrar cada línea de texto
        print(f"Resultados de la ruta empezando en: {ciudadOrigen}".center(anchoTabla))
        print(f"Distancia total recorrida: {distanciaTotal} km".center(anchoTabla))
        print("\n" + "<< Desglose de la ruta optimizada >>".center(anchoTabla))
        print(tablaTexto)
        print("-" * anchoTabla) # Ajusto también la línea de separación al ancho de la tabla
        print("\n")
    
    return dfResultadoRuta

# ¿Desde qué ciudad la ruta es más corta?
# Calculo el kilometraje total para cada ciudad de la matriz para encontrar el origen óptimo sin generar ruido por pantalla
resultadosKilometraje = {}
for ciudad in dfDistanciasLimpio.index:
    # Obtengo el DataFrame para comparar las distancias acumuladas finales
    dfTemporal = calcularRutaMasCorta(dfDistanciasLimpio, ciudad, mostrarResultados=False)
    resultadosKilometraje[ciudad] = dfTemporal["Distancia Acumulada (km)"].iloc[-1]

# Determino cuál es la mejor ciudad comparando los resultados de mi diccionario
mejorCiudadOrigen = min(resultadosKilometraje, key=resultadosKilometraje.get)
```

### Llamada de la función

```python
# Ejecuto la llamada definitiva para la mejor ciudad encontrada; la propia función se encargará de imprimir todo
dfMejorRuta = calcularRutaMasCorta(dfDistanciasLimpio, mejorCiudadOrigen)
```
