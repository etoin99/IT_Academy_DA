### Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. 
Pots obtenir més informació del seu càlcul en: 🔗 Índice de masa corporal IMC que es y como se calcula.

### La funció ha de classificar el resultat en les seves respectives categories.

### Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.

### Código de la función

```python
# El cálculo de (Peso * Altura² "al cuadrado"), me devuelve el Índice de Masa Corporal (IMC).
def calculadoraIMC(peso, altura):
    return round(peso / altura ** 2, 2)

# Una vez sabemos el IMC, utilizo IF para que compare el valor y en función del resultado, devuelva el Estado Corporal con RETURN.
def IMCDefinicion(defIMC):
    if defIMC > 30:
        return "Obesidad"
    elif defIMC >= 25: 
        return "Sobrepeso"
    elif defIMC >=18.5:
        return "Peso normal"
    else:
        return "Bajo Peso"
    

```

## 2. Convertidor de temperatures

### Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

### Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).

### Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduides per l’usuari/a no estan en el format adequat.

### (EXTRA): Pensa una manera d’emmagatzemar totes les possibles conversions en un sol objecte (Llista? Diccionari? DataFrame?) en comptes d’escriure molts if else en funció de la temperatura d’origen i la temperatura de destí.

### Código de la función

```python
import pandas as pd
dfConversor = pd.read_excel("ConversorTemperatura.xlsx", index_col = 0)
```

```python
dfConversor
```

```python
def ConvertidorTemperatura(origen, destino, temperatura, dfConversor=dfConversor):
    origen = origen.strip().upper()
    destino = destino.strip().upper()
    
    # Validación de temperatura
    try:
        temperatura = float(temperatura.strip())
    except ValueError:
        return "Error: La temperatura debe ser un número válido (ej. 25 o 25.5)."
    
    # Validación de unidades en el DataFrame
    if origen not in dfConversor.index or destino not in dfConversor.columns:
        unidades_validas = ["1. °C (Celsius)", "2. °F (Fahrenheit)", "3. K (Kelvin)", "4. °Ré (Reaumur)", "5. °Ra (Rankine)"]
        return f"Error: Unidad '{origen}' o '{destino}' no válida. Escribe una de las siguientes: {', '.join(unidades_validas)}"
    
    # Utilizo eval para ejecutar la fórmula tipo string del Excel
    formula = dfConversor.loc[origen, destino]
    resultado = eval(formula, {"__builtins__": {}}, {"x": temperatura})
    resultado_redondeado = round(resultado, 2)
    

    # Lo guardo en un diccionario
    return {
        "Unidad de origen": origen,
        "Temperatura de origen": temperatura,
        "Unidad de destino": destino,
        "Temperatura de destino": resultado_redondeado
    }
```

### Ejecución

```python
# Guardo el resultado en una variable
resultado = ConvertidorTemperatura(
    input("Unidad de origen (ej. Celsius, Fahrenheit, Kelvin, Reaumur o Rankine): "), 
    input("Unidad de destino (ej. Celsius, Fahrenheit, Kelvin, Reaumur o Rankine): "), 
    input("Temperatura: "))

dfResultado = pd.DataFrame([resultado])

dfResultado
```

## 3. Comptador de paraules d'un text.

### Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. 
### Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

### (EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? 
### “Hola com va?” hauria de retornar (4+3+2) / 3 = 3

### Código de la función

```python
import statistics
```

```python
def ContadorPalabras(ListaTXT1):
    palabras = ListaTXT1.lower().split()
    conteo = {}
    longitudes = []

    for palabra in palabras:
        palabra = palabra.strip('.,;:!?()"')
        conteo[palabra] = conteo.get(palabra, 0) + 1
        longitudes.append(len(palabra))
        
    MediaLargo = round(statistics.mean(longitudes), 2)
    
    return {
        "Conteo de palabras" : conteo,
        "Media del largo" : MediaLargo}
```

## 4. Diccionari invers (amb possibilitat de duplicats)

### Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, 
### intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, 
### la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

### Código de la función

```python
def InvertirDiccionario(DiccionarioNoInvertido):
    DiccionarioInvertido = {}

    for clave, valor in DiccionarioNoInvertido.items():
        # Si el valor ya existe en el nuevo diccionario, se gestiona el duplicado
        if valor in DiccionarioInvertido:
            print(f"Advertencia: El valor '{valor}' ja existe. Añadiré la clave repetida.")
            
            # Si todavía no es una lista (primer duplicado que encontramos)
            if not isinstance(DiccionarioInvertido[valor], list):
                DiccionarioInvertido[valor] = [DiccionarioInvertido[valor], clave]
            else:
                # Si ya es una lista, solo añade la nueva clave
                DiccionarioInvertido[valor].append(clave)
        else:
            # Si es la primera vez que ve el valor, lo añade
            DiccionarioInvertido[valor] = clave
    
    return DiccionarioInvertido
```

### Ejecución

```python
DiccionarioParaInvertir = ({'a':1, 'b':2, 'c':3})

ResultadoDiccionarioInvertido = InvertirDiccionario(DiccionarioParaInvertir)
print(ResultadoDiccionarioInvertido)
```

```python
# Prueba con duplicados
prueba = {'x': 'apple', 'y': 'banana', 'z': 'banana'}
ResultadoDuplicados = InvertirDiccionario(prueba)
print(ResultadoDuplicados)
```

## 1. Comptador i endreçador de paraules d'un text.

### El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT;
### i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, 
### és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z. 

### Código de la función

```python
import re
import pprint
# Limpieza del texto
def LimpiadorDeTexto(archivo):
    TextoLimpioFun = []
    # Uso un bucle para leer línea por línea del archivo
    for linea in archivo:
        # Separo por espacios para sacar las palabras
        palabraFrase = linea.split()
        for letras in palabraFrase:
            # Quito símbolos raros con re.sub (solo dejo letras y números)
            palabra = re.sub('[^a-zA-Z0-9áéíóúÁÉÍÓÚñÑüÜ]', '', letras)
            # Paso a minúscula y compruebo que no sea un espacio vacío
            if palabra != "":
                TextoLimpioFun.append(palabra.lower())
    
    return TextoLimpioFun

# Organizo por inicial
def AgrupadorPorInicial(ListaPalabras):
    # Aquí creo el "diccionario de diccionarios"
    # Estructura: {'a': {'agua': 1}, 'b': {'blanca': 2}}
    resultado = {}

    for palabra in ListaPalabras:
        inicial = palabra[0] # Cojo la primera letra
        
        # Si la letra no está en el diccionario grande, la añado con un sub-diccionario vacío
        if inicial not in resultado:
            resultado[inicial] = {}
            
        # Una vez dentro del sub-diccionario de esa letra, cuento la palabra
        if palabra in resultado[inicial]:
            resultado[inicial][palabra] += 1 # Si ya existe, sumo uno
        else:
            resultado[inicial][palabra] = 1 # Si es nueva para esta letra, empieza en 1
            
    return resultado
```

### Ejecución

```python
# Ejecución y Ordenación
with open("tu_me_quieres_blanca.txt", encoding="utf-8") as archivo:
    lista_limpia = LimpiadorDeTexto(archivo)
    datos_agrupados = AgrupadorPorInicial(lista_limpia)

# Para ordenar convierto el diccionario en una lista de tuplas y ordeno por la inicial
# También ordeno los sub-diccionarios internos
DiccionarioOrdenado = {}
for letra in sorted(datos_agrupados.keys()):
    # Ordeno las palabras dentro de cada letra
    palabras_internas = dict(sorted(datos_agrupados[letra].items()))
    DiccionarioOrdenado[letra] = palabras_internas

# Muestro el resultado final usando la librería pprint
pprint.pprint(DiccionarioOrdenado)
```

## 2. Conversió de tipus de dades.

### El client rep una llista de dades i necessita generar dues llistes, 
### la primera on estaran tots els elements que es van poder convertir en flotants 
### i l'altra on estan els elements que no es van poder convertir. Exemple de la llista que rep el client:

### Código de la función

```python
def FiltrarDatos(ListaEntrada):
    Convertibles = []
    NoConvertibles = []
    
    # Creo una lista plana para no liarme con las sublistas o tuplas
    ElementosSueltos = []
    for item in ListaEntrada:
        if isinstance(item, (list, tuple, set)):
            # Si es una lista o tupla, meto sus elementos uno a uno
            ElementosSueltos.extend(item)
        else:
            ElementosSueltos.append(item)
            
    # Ahora recorro cada elemento e intento la conversión
    for elemento in ElementosSueltos:
        try:
            # Intento transformarlo en un número decimal
            numero = float(elemento)
            Convertibles.append(numero)
        except (ValueError, TypeError):
            # Si da error (porque es una palabra o un formato raro), 
            # lo guardo en la otra lista
            NoConvertibles.append(elemento)
            
    return Convertibles, NoConvertibles
```

### Ejecución

```python
# Datos del ejemplo
DatosCliente = [ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]

# Ejecuto la función y guardo el resultado
ListaConvertibles, ListaFallos = FiltrarDatos(DatosCliente)

print(f"Números convertidos: {ListaConvertibles}")
print(f"Cosas que no son números: {ListaFallos}")
```

## 1. Generador de Contrasenyes

### Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

### En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. 
### Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. 
### La funció ha de dependre dels següents paràmetres:


* longitud (int): Longitud de la contrasenya
* majuscules (bool = True): Si hi ha d’aparèixer majúscules
* minuscules (bool = True): Si hi ha d’aparèixer minúscules
* numeros (bool = True): Si hi ha d’aparèixer números
* signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)

### Així doncs, si executem la funció de la següent manera:

crear_contrasenya(10, True, True, True, True)

### Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:

9Er,5Vn8P$

### Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

### (EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).

### Código de la función

```python
import numpy as np
import string # Para acceder a los caracteres alfabéticos y numéricos
import pyperclip # Para conseguir el EXTRA del enunciado y que se copie al portapapeles

def GenerarContraseña(longitud, majuscules=True, minuscules=True, numeros=True, signes=False):
    # Creo una "bolsa" vacía donde meteremos todos los caracteres permitidos
    CaracteresPosibles = ""
    
    # Creo una lista para asegurar al menos uno de cada
    Garantizados = []
    
    # Reviso los parámetros y añado a la bolsa
    if minuscules:
        CaracteresPosibles += string.ascii_lowercase  # letras a-z
        Garantizados.append(np.random.choice(list(string.ascii_lowercase))) # Guardo uno de este tipo

        
    if majuscules:
        CaracteresPosibles += string.ascii_uppercase  # letras A-Z
        Garantizados.append(np.random.choice(list(string.ascii_uppercase))) # Guardo uno de este tipo
        
    if numeros:
        CaracteresPosibles += string.digits           # números 0-9
        Garantizados.append(np.random.choice(list(string.digits))) # Guardo uno de este tipo
        
    if signes:
        # Aquí defino los símbolos que quiero usar
        simbolos = "!#$%&'()*+,-./:;<=>?@[\]^_{|}~`"
        CaracteresPosibles += simbolos 
        Garantizados.append(np.random.choice(list(simbolos))) # Guardo uno de este tipo

    # Validación: si el usuario pone todo en False, no podemos generar nada
    if not CaracteresPosibles:
        return "Error: Debes seleccionar al menos un tipo de carácter"
    
    # Si la longitud es pequeña (ejemplo 4) y hay 4 tipos, la lista "Garantizados" ya tiene todo.
    # Si la longitud es mayor, rellena con lo que falta.
    if longitud > len(Garantizados):
        faltan = longitud - len(Garantizados)
        lista_caracteres = list(CaracteresPosibles)
        relleno = np.random.choice(lista_caracteres, size=faltan, replace=True)
        seleccion_final = Garantizados + list(relleno)
    else:
        # Si el usuario pide una contraseña con solamente 2 caracteres pero activa los 4 tipos como TRUE, corta para respetar la "longitud"
        seleccion_final = Garantizados[:longitud]

    # Mezclo la selección para que no siempre empiece por el mismo orden
    np.random.shuffle(seleccion_final)

    # Uno la lista devuelta en una sola palabra (la contraseña)
    ContraseñaFinal = "".join(seleccion_final)
    
    pyperclip.copy(ContraseñaFinal) # Extra del portapapeles

    return ContraseñaFinal 
```

### Ejecución

```python
# Ejemplo de uso con una longitud de 16 carácteres,
# ya que 16 caracteres o más son ideales según las últimas directrices del NIST y expertos en ciberseguridad.
Contraseña = GenerarContraseña(16, True, True, True, True)
print(f"Tu nueva contraseña es: {Contraseña}")
print("Ha sido copiada en el portapapeles.")
```

## 2. Processament de dades simple

### Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. 
### Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. 
### Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.

* Utilitza el fitxer "historic_partits.txt"

### Necessita un programa que retorni:

* El nombre total de gols que ha fet cada equip.
* El nom de l’equip més golejador.
* El nom de l’equip més golejat
* La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)

### Código de la función

```python
# Creo la función para procesar el archivo (Añado contadores de victorias/empates/derrotas)
def ProcesarLiga(NombreArchivo):
    equipos = {}

    with open(NombreArchivo, "r", encoding="utf-8") as archivo:
        for linea in archivo:
            partes = linea.strip().split('\t')
            if len(partes) < 3:
                continue 
                
            EquipoLocal = partes[0]
            Resultado = partes[1]
            EquipoVisitante = partes[2]
            
            GolesLocal, GolesVisitante = map(int, Resultado.split('-'))
            
            # Inicializo con todas las columnas que muestra el ejemplo del ejercicio
            for equipo in [EquipoLocal, EquipoVisitante]:
                if equipo not in equipos:
                    equipos[equipo] = {
                        'partits guanyats': 0, 
                        'partits empatats': 0, 
                        'partits perduts': 0, 
                        'gols': 0, 
                        'gols encaixats': 0, 
                        'punts': 0
                    }

            # Sumo goles ("golesF; Goles Favorables" es 'gols' y "golesC; Goles en contra" es 'gols encaixats')
            equipos[EquipoLocal]['gols'] += GolesLocal
            equipos[EquipoLocal]['gols encaixats'] += GolesVisitante
            equipos[EquipoVisitante]['gols'] += GolesVisitante
            equipos[EquipoVisitante]['gols encaixats'] += GolesLocal

            # Lógica de puntos y partidos
            if GolesLocal > GolesVisitante:
                equipos[EquipoLocal]['punts'] += 3
                equipos[EquipoLocal]['partits guanyats'] += 1
                equipos[EquipoVisitante]['partits perduts'] += 1
            elif GolesLocal < GolesVisitante:
                equipos[EquipoVisitante]['punts'] += 3
                equipos[EquipoVisitante]['partits guanyats'] += 1
                equipos[EquipoLocal]['partits perduts'] += 1
            else:
                equipos[EquipoLocal]['punts'] += 1
                equipos[EquipoVisitante]['punts'] += 1
                equipos[EquipoLocal]['partits empatats'] += 1
                equipos[EquipoVisitante]['partits empatats'] += 1
                
    return equipos
```

### Ejecución

```python
# Genero el Diccionario
DatosLiga = ProcesarLiga("historic_partits.txt")

# En este punto procedo a convertir a DataFrame
# Creo el DataFrame; "orient=index" hace que las claves (Equipos) sean las filas.
dfResultadoLiga = pd.DataFrame.from_dict(DatosLiga, orient='index')

# Ordeno las columnas para que salgan exactamente en el orden que aparece en el ejemplo.
ColumnasOrdenadas = [
    'partits guanyats', 
    'partits empatats', 
    'partits perduts', 
    'gols', 
    'gols encaixats', 
    'punts'
]
dfResultadoLiga = dfResultadoLiga[ColumnasOrdenadas]

# Ordeno la tabla por puntos (de más a menos) para que sea una clasificación real
dfResultadoLiga = dfResultadoLiga.sort_values(by='punts', ascending=False)

# Muestro el resultado final
dfResultadoLiga
```

```python
# Resultados específicos del ejercicio

# 1. El nombre total de gols que ha fet cada equip.
# Muestro la columna 'gols' de nuestro DataFrame
GolesTotales = dfResultadoLiga[['gols']] 

# 2. El nom de l’equip més golejador. (el que tiene más goles a favor)
# .idxmax() nos da el nombre del equipo con el valor más alto
EquipoMaxGoleador = dfResultadoLiga['gols'].idxmax()
CantidadMaxGoleador = dfResultadoLiga['gols'].max()

# 3. El nom de l’equip més golejat (el que tiene más goles en contra)
EquipoMaxGoleado = dfResultadoLiga['gols encaixats'].idxmax()
CantidadMaxGoleado = dfResultadoLiga['gols encaixats'].max()

# Muestro los resultados con un print
print("\n" + " DESTACADOS DE LA TEMPORADA ".center(60, "="))
print(f" Equipo más goleador: {EquipoMaxGoleador} con {CantidadMaxGoleador} goles.")
print(f" Equipo más goleado: {EquipoMaxGoleado} con {CantidadMaxGoleado} goles recibidos.")
print("-" * 60)
print("\n Goles totales por equipo:")
GolesTotales
```
